## Chi-Square Test Code

```python
from scipy.stats import chi2_contingency

categorical_features = [
    'Gender',
    'Smoking',
    'GeneticRisk',
    'CancerHistory'
]

for feature in categorical_features:

    contingency_table = pd.crosstab(
        df[feature],
        df['Diagnosis']
    )

    chi2, p_value, dof, expected = chi2_contingency(
        contingency_table
    )

    print("\nFeature:", feature)
    print("Chi-Square:", round(chi2, 4))
    print("P-Value:", round(p_value, 6))
    print("Degrees of Freedom:", dof)
```

## V Test

```python
import numpy as np
import pandas as pd
from scipy.stats import chi2_contingency

categorical_features = [
    'Gender',
    'Smoking',
    'GeneticRisk',
    'CancerHistory'
]

for feature in categorical_features:

    table = pd.crosstab(df[feature], df['Diagnosis'])

    chi2, p_value, dof, expected = chi2_contingency(table)

    n = table.to_numpy().sum()

    min_dimension = min(table.shape[0] - 1, table.shape[1] - 1)

    cramer_v = np.sqrt(chi2 / (n * min_dimension))

    print("\nFeature:", feature)
    print("Cramer's V:", round(cramer_v, 4))
    print("Minimum Expected Frequency:", round(expected.min(), 2))
```
